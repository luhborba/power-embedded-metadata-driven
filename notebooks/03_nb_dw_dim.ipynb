```python
# =============================================================================
# nb_dw_dim — Worker DW genérico para dimensões
# Chamado via runMultiple pelo nb_orquestrador
# Lê da ODS, aplica taxonomia de nomes, deduplica e grava na camada DW
#
# Válido para: dw_dim_dataset, dw_dim_report, dw_dim_user,
#              dw_dim_group, dw_dim_folder
# =============================================================================
# CÉLULA 1 — Parâmetros (tag esta célula como "parameters" no Fabric)
# =============================================================================

if 'dw_table_name'    not in globals(): dw_table_name    = "dw_dim_dataset"  # ex: "dw_dim_user"
if 'source_ods_table' not in globals(): source_ods_table = "ods_datasets"  # ex: "ods_user"
```

```python
# =============================================================================
# CÉLULA 2 — Helpers e configuração Spark
# =============================================================================
```

```python
logger = setup_logger("nb_dw_dim")
configure_spark_dw()
```

```python
# =============================================================================
# CÉLULA 3 — Mapeamento de colunas por dimensão (taxonomia DW)
#
# Prefixos:
#   sk_  → surrogate key (chave da dimensão)
#   nk_  → natural key (ID original da fonte)
#   fk_  → foreign key (referência a outra dimensão)
#   nm_  → nome / descrição curta
#   ds_  → descrição longa
#   cd_  → código / tipo
#   fl_  → flag booleano
#   dt_  → data / timestamp
#   url_ → URL
# =============================================================================

DIM_CONFIG = {

    "dw_dim_dataset": {
        "rename": {
            "id":                             "sk_dataset",
            "organizationId":                 "cd_organizacao",
            "workspaceId":                    "cd_workspace",
            "datasetId":                      "nk_dataset_pbi",
            "datasetName":                    "nm_dataset",
            "requiresEffectiveIdentity":      "fl_requer_identidade_efetiva",
            "requiresEffectiveIdentityRoles": "fl_requer_roles_identidade",
            "requiresOnPremisesGateway":      "fl_requer_gateway_on_prem",
            "selectedRls":                    "ds_rls_selecionado",
        },
        "drop": ["dt_extracao"],
    },

    "dw_dim_report": {
        "rename": {
            "id":            "sk_report",
            "name":          "nm_report",
            "description":   "ds_report",
            "datasetId":     "fk_dataset",
            "embedUrl":      "url_embed",
            "workspaceId":   "cd_workspace",
            "reportType":    "cd_tipo_report",
            "folderId":      "fk_folder",
            "workspaceName": "nm_workspace",
            "userOwner":     "nm_usuario_proprietario",
        },
        "drop": ["dt_extracao"],
    },

    "dw_dim_user": {
        "rename": {
            "id":                              "sk_user",
            "email":                           "nm_email",
            "name":                            "nm_usuario",
            "role":                            "cd_perfil",
            "reportLandingPage":               "ds_pagina_inicial",
            "department":                      "nm_departamento",
            "expirationDate":                  "dt_expiracao",
            "windowsAdUser":                   "nm_usuario_ad",
            "canEditReport":                   "fl_pode_editar_report",
            "canCreateReport":                 "fl_pode_criar_report",
            "canOverwriteReport":              "fl_pode_sobrescrever_report",
            "canRefreshDataset":               "fl_pode_atualizar_dataset",
            "canCreateSubscription":           "fl_pode_criar_assinatura",
            "canDownloadPbix":                 "fl_pode_baixar_pbix",
            "canExportReport":                 "fl_pode_exportar_report",
            "canExportReportWithHiddenPages":  "fl_pode_exportar_paginas_ocultas",
            "hideHomePages":                   "fl_ocultar_pagina_inicial",
            "hideAddressBook":                 "fl_ocultar_lista_contatos",
            "hideFavorites":                   "fl_ocultar_favoritos",
            "hideMostAccessed":                "fl_ocultar_mais_acessados",
            "hideRecentlyAccessed":            "fl_ocultar_recentes",
            "hideRecentlyPublished":           "fl_ocultar_publicados_recentes",
            "hideFolders":                     "fl_ocultar_pastas",
            "canCreateNewUsers":               "fl_pode_criar_usuarios",
            "canStartCapacityByDemand":        "fl_pode_iniciar_capacidade",
            "canDisplayVisualHeaders":         "fl_pode_exibir_cabecalhos",
            "accessReportAnyTime":             "fl_acesso_report_qualquer_hora",
        },
        "drop": ["dt_extracao", "reports", "companies", "groups", "reportNames"],
    },

    "dw_dim_group": {
        "rename": {
            "id":                   "sk_group",
            "organizationId":       "cd_organizacao",
            "name":                 "nm_grupo",
            "description":         "ds_grupo",
            "canEditReport":        "fl_pode_editar_report",
            "canCreateReport":      "fl_pode_criar_report",
            "canOverwriteReport":   "fl_pode_sobrescrever_report",
            "canRefreshDataset":    "fl_pode_atualizar_dataset",
            "bypassFirewall":       "fl_bypass_firewall",
        },
        "drop": ["dt_extracao", "reports", "users"],
    },

    "dw_dim_folder": {
        "rename": {
            "id":             "sk_folder",
            "organizationId": "cd_organizacao",
            "name":           "nm_pasta",
            "parentFolderId": "fk_pasta_pai",
        },
        "drop": ["dt_extracao"],
    },
}
```

```python
# =============================================================================
# CÉLULA 4 — Validar configuração
# =============================================================================

if dw_table_name not in DIM_CONFIG:
    raise ValueError(f"dw_table_name '{dw_table_name}' não encontrado em DIM_CONFIG. "
                     f"Valores válidos: {list(DIM_CONFIG.keys())}")

config = DIM_CONFIG[dw_table_name]

LAKEHOUSE_BRONZE  = "lh_bronze"
LAKEHOUSE_GOLD  = "lh_gold"
ODS_SCHEMA = "power_embedded"
DW_SCHEMA  = "power_embedded"

workspace = mssparkutils.env.getWorkspaceName()

ods_table_path = build_abfss_path(workspace, LAKEHOUSE_BRONZE, "Tables", ODS_SCHEMA, source_ods_table)
dw_table_path  = build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA,  dw_table_name)

logger.info(f"dw_table_name   : {dw_table_name}")
logger.info(f"source_ods_table: {source_ods_table}")
```

```python
# =============================================================================
# CÉLULA 5 — Leitura da ODS
# =============================================================================

df = spark.read.format("delta").load(ods_table_path)
logger.info(f"{df.count()} registro(s) lidos | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 6 — Remover colunas desnecessárias
# =============================================================================

from pyspark.sql import functions as F
from pyspark.sql.types import TimestampType
from datetime import datetime, timezone

cols_to_drop = [c for c in config["drop"] + ["dt_carga"] if c in df.columns]
df = df.drop(*cols_to_drop)
logger.info(f"Colunas removidas: {cols_to_drop}")
```

```python
# =============================================================================
# CÉLULA 7 — Aplicar taxonomia (rename de colunas)
# =============================================================================

for old_name, new_name in config["rename"].items():
    if old_name in df.columns:
        df = df.withColumnRenamed(old_name, new_name)

logger.info(f"Taxonomia aplicada | colunas finais: {df.columns}")
```

```python
# =============================================================================
# CÉLULA 8 — Deduplicar
# =============================================================================

before = df.count()
df = df.dropDuplicates()
after  = df.count()

if before != after:
    logger.info(f"Deduplicação: {before - after} linha(s) removida(s)")
```

```python
# =============================================================================
# CÉLULA 9 — Adicionar dt_carga
# =============================================================================

dt_carga = datetime.now(timezone.utc)
df = df.withColumn("dt_carga", F.lit(dt_carga).cast(TimestampType()))
```

```python
# =============================================================================
# CÉLULA 10 — Gravar DW (overwrite total — dims são sempre full load)
# =============================================================================

logger.info(f"Gravando {dw_table_name} — mode: overwrite")

(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .save(dw_table_path)
)

row_count = df.count()
logger.info(f"DW gravado: {dw_table_name} | {row_count} linha(s) | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 11 — Finalização
# =============================================================================

try:
    log_execucao(logger, "DW", dw_table_name, "success", row_count)
except Exception as e:
    logger.warning(f"Falha ao registrar execução (não crítico): {e}")

result = {
    "status":           "success",
    "dw_table_name":    dw_table_name,
    "source_ods_table": source_ods_table,
    "rows":             row_count,
    "columns":          len(df.columns),
}

logger.info(f"nb_dw_dim finalizado | {result}")
mssparkutils.notebook.exit(json.dumps(result))
```
