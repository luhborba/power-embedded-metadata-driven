```python
# =============================================================================
# nb_dw_fact_email_audit — Fato DW: Auditoria de E-mails
# Fonte: ods_email_audit
# Destino: dw_fact_email_audit
#
# Cada linha = um e-mail enviado pelo Power Embedded (notificações, alertas).
# Fato isolado — sem referência a dimensões.
#
# Schema ODS confirmado:
#   id, organizationId, createdAt, templateName, subject, companyName,
#   destination, smtpHost, status, error, dt_extracao, dt_carga
# =============================================================================
# CÉLULA 1 — Parâmetros (tag esta célula como "parameters" no Fabric)
# =============================================================================

if 'dw_table_name'    not in globals(): dw_table_name    = "dw_fact_email_audit"
if 'source_ods_table' not in globals(): source_ods_table = "power_embedded_email_audit"
```

```python
# =============================================================================
# CÉLULA 2 — Helpers e configuração Spark
# =============================================================================
```

```python
%run 00_nb_helpers
```

```python
logger = setup_logger("nb_dw_fact_email_audit")
configure_spark_dw()
```

```python
# =============================================================================
# CÉLULA 3 — Constantes e caminhos
# =============================================================================

from pyspark.sql import functions as F
from pyspark.sql.types import TimestampType
from datetime import datetime, timezone

LAKEHOUSE_BRONZE  = "lh_bronze"
LAKEHOUSE_GOLD  = "lh_gold"
ODS_SCHEMA = "power_embedded"
DW_SCHEMA  = "power_embedded"

workspace = mssparkutils.env.getWorkspaceName()

ods_table_path = build_abfss_path(workspace, LAKEHOUSE_BRONZE, "Tables", ODS_SCHEMA, source_ods_table)
dw_table_path  = build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA,  dw_table_name)

logger.info(f"source_ods_table: {source_ods_table}")
logger.info(f"dw_table_name   : {dw_table_name}")
```

```python
# =============================================================================
# CÉLULA 4 — Leitura da ODS
# =============================================================================

df = spark.read.format("delta").load(ods_table_path)
logger.info(f"{df.count()} e-mail(s) lidos | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 5 — Aplicar taxonomia
# =============================================================================

df = df.drop("dt_extracao", "dt_carga")

RENAME_MAP = {
    "id":             "nk_email_audit",  # natural key do evento de e-mail
    "organizationId": "cd_organizacao",
    "createdAt":      "dt_envio",        # timestamp do envio
    "templateName":   "nm_template",     # template de e-mail utilizado
    "subject":        "ds_assunto",      # assunto do e-mail
    "companyName":    "nm_empresa",      # empresa destinatária
    "destination":    "nm_destino",      # endereço de destino
    "smtpHost":       "nm_smtp_host",    # servidor SMTP utilizado
    "status":         "cd_status",       # código de status do envio
    "error":          "ds_erro",         # mensagem de erro (null se sucesso)
}

for old_name, new_name in RENAME_MAP.items():
    if old_name in df.columns:
        df = df.withColumnRenamed(old_name, new_name)

logger.info(f"Taxonomia aplicada | colunas: {df.columns}")
```

```python
# =============================================================================
# CÉLULA 6 — Deduplicar por natural key
# =============================================================================

before = df.count()
df = df.dropDuplicates(["nk_email_audit"])
after  = df.count()

if before != after:
    logger.info(f"Deduplicação: {before - after} linha(s) removida(s)")
```

```python
# =============================================================================
# CÉLULA 7 — Adicionar dt_carga
# =============================================================================

dt_carga = datetime.now(timezone.utc)
df = df.withColumn("dt_carga", F.lit(dt_carga).cast(TimestampType()))
```

```python
# =============================================================================
# CÉLULA 8 — Gravar DW
# =============================================================================

logger.info(f"Gravando {dw_table_name} — mode: overwrite")

(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .save(dw_table_path)
)

row_count = df.count()
logger.info(f"DW gravado: {dw_table_name} | {row_count} linha(s) | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 9 — Finalização
# =============================================================================

result = {
    "status":           "success",
    "dw_table_name":    dw_table_name,
    "source_ods_table": source_ods_table,
    "rows":             row_count,
    "columns":          len(df.columns),
}

logger.info(f"nb_dw_fact_email_audit finalizado | {result}")
try:
    log_execucao(logger, "DW", dw_table_name, "success", row_count)
except Exception as e:
    logger.warning(f"Falha ao registrar execução (não crítico): {e}")

mssparkutils.notebook.exit(json.dumps(result))
```
