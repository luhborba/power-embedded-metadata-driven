```python
# =============================================================================
# nb_dw_bridge_permission — Bridge DW: Permissões
# Fonte: ods_permission_report
# Destino: dw_bridge_permission
#
# Resolve o relacionamento N:N entre relatórios e usuários/grupos via RLS.
# Cada linha = uma permissão de um usuário (via grupo) em um relatório.
#
# Schema ODS confirmado:
#   organizationId, type, userEmail, reportName, workspaceName,
#   permissionDetail, rlsByGroup, dt_extracao, dt_carga
# =============================================================================
# CÉLULA 1 — Parâmetros (tag esta célula como "parameters" no Fabric)
# =============================================================================

if 'dw_table_name'    not in globals(): dw_table_name    = "dw_bridge_permission"
if 'source_ods_table' not in globals(): source_ods_table = "power_embedded_permission_report"
```

```python
# =============================================================================
# CÉLULA 2 — Helpers e configuração Spark
# =============================================================================
```

```python
%run 00_nb_helpers
```

```python
logger = setup_logger("nb_dw_bridge_permission")
configure_spark_dw()
```

```python
# =============================================================================
# CÉLULA 3 — Constantes e caminhos
# =============================================================================

from pyspark.sql import functions as F
from pyspark.sql.types import TimestampType
from datetime import datetime, timezone

LAKEHOUSE_BRONZE  = "lh_bronze"
LAKEHOUSE_GOLD  = "lh_gold"
ODS_SCHEMA = "power_embedded"
DW_SCHEMA  = "power_embedded"

workspace = mssparkutils.env.getWorkspaceName()

ods_table_path = build_abfss_path(workspace, LAKEHOUSE_BRONZE, "Tables", ODS_SCHEMA, source_ods_table)
dw_table_path  = build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA,  dw_table_name)

logger.info(f"source_ods_table: {source_ods_table}")
logger.info(f"dw_table_name   : {dw_table_name}")
```

```python
# =============================================================================
# CÉLULA 4 — Leitura da ODS
# =============================================================================

df = spark.read.format("delta").load(ods_table_path)
logger.info(f"{df.count()} registro(s) lidos | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 5 — Remover colunas técnicas + aplicar taxonomia
# =============================================================================

cols_to_drop = [c for c in ["dt_extracao", "dt_carga"] if c in df.columns]
df = df.drop(*cols_to_drop)

RENAME_MAP = {
    "organizationId":   "cd_organizacao",
    "type":             "cd_tipo_permissao",     # ex: "GroupRlsPermission"
    "userEmail":        "nm_email",
    "reportName":       "nm_report",
    "workspaceName":    "nm_workspace",
    "permissionDetail": "ds_detalhe_permissao",  # ex: "Role: Geral"
    "rlsByGroup":       "nm_grupo_rls_raw",       # ex: "Group: # Gestores Regionais PA" — limpado abaixo
}

for old_name, new_name in RENAME_MAP.items():
    if old_name in df.columns:
        df = df.withColumnRenamed(old_name, new_name)

# Remove prefixo "Group: " do campo de grupo RLS para permitir join com dw_dim_group
# "Group: # Gestores Regionais PA" → "# Gestores Regionais PA"
if "nm_grupo_rls_raw" in df.columns:
    df = df.withColumn(
        "nm_grupo_rls",
        F.regexp_replace(F.col("nm_grupo_rls_raw"), r"^Group:\s*", "")
    ).drop("nm_grupo_rls_raw")

# Remove prefixo "Role: " do campo de detalhe
# "Role: Geral" → "Geral"
if "ds_detalhe_permissao" in df.columns:
    df = df.withColumn(
        "nm_role_rls",
        F.regexp_replace(F.col("ds_detalhe_permissao"), r"^Role:\s*", "")
    ).drop("ds_detalhe_permissao")

logger.info(f"Taxonomia aplicada | colunas: {df.columns}")
```

```python
# =============================================================================
# CÉLULA 6 — Enriquecer com surrogate keys das dimensões
#
# A ODS não traz IDs — apenas nomes. Fazemos lookup nas dims para trazer
# os surrogate keys e garantir integridade referencial no modelo analítico.
# Joins left: linhas sem match ficam com fk null (investigar após carga).
# Leitura via ABFSS path (não spark.table) para evitar dependência do lakehouse default.
# =============================================================================

dim_user   = spark.read.format("delta").load(build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA, "dw_dim_user")).select("nm_email", "sk_user")
dim_report = spark.read.format("delta").load(build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA, "dw_dim_report")).select("nm_report", "nm_workspace", "sk_report")
dim_group  = spark.read.format("delta").load(build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA, "dw_dim_group")).select("nm_grupo", "sk_group")

# Join usuário por e-mail
df = df.join(dim_user, on="nm_email", how="left")

# Join relatório por nome + workspace (evita ambiguidade entre workspaces)
df = df.join(dim_report, on=["nm_report", "nm_workspace"], how="left")

# Join grupo pelo nome limpo
df = df.join(
    dim_group.withColumnRenamed("nm_grupo", "nm_grupo_rls"),
    on="nm_grupo_rls",
    how="left"
)

# Verificar cobertura dos joins
nulls_user   = df.filter(F.col("sk_user").isNull()).count()
nulls_report = df.filter(F.col("sk_report").isNull()).count()
# Conta apenas grupos onde havia valor mas não encontrou match — null é esperado (sem RLS de grupo)
nulls_group  = df.filter(F.col("sk_group").isNull() & F.col("nm_grupo_rls").isNotNull()).count()

total = df.count()
logger.info(f"Join cobertura | total: {total} | sk_user null: {nulls_user} | sk_report null: {nulls_report} | sk_group sem match: {nulls_group}")

if nulls_user > 0:
    logger.warning(f"{nulls_user} linha(s) sem match em dw_dim_user — verificar nm_email")
if nulls_report > 0:
    logger.warning(f"{nulls_report} linha(s) sem match em dw_dim_report — verificar nm_report/nm_workspace")
if nulls_group > 0:
    logger.warning(f"{nulls_group} linha(s) com nm_grupo_rls sem match em dw_dim_group — verificar nomes")
```

```python
# =============================================================================
# CÉLULA 7 — Deduplicar
# =============================================================================

before = df.count()
df = df.dropDuplicates()
after  = df.count()

if before != after:
    logger.info(f"Deduplicação: {before - after} linha(s) removida(s)")
```

```python
# =============================================================================
# CÉLULA 8 — Adicionar dt_carga
# =============================================================================

dt_carga = datetime.now(timezone.utc)
df = df.withColumn("dt_carga", F.lit(dt_carga).cast(TimestampType()))
```

```python
# =============================================================================
# CÉLULA 9 — Gravar DW
# =============================================================================

logger.info(f"Gravando {dw_table_name} — mode: overwrite")

(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .save(dw_table_path)
)

row_count = df.count()
logger.info(f"DW gravado: {dw_table_name} | {row_count} linha(s) | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 10 — Finalização
# =============================================================================

result = {
    "status":           "success",
    "dw_table_name":    dw_table_name,
    "source_ods_table": source_ods_table,
    "rows":             row_count,
    "columns":          len(df.columns),
}

logger.info(f"nb_dw_bridge_permission finalizado | {result}")
try:
    log_execucao(logger, "DW", dw_table_name, "success", row_count)
except Exception as e:
    logger.warning(f"Falha ao registrar execução (não crítico): {e}")

mssparkutils.notebook.exit(json.dumps(result))
```
