```python
# =============================================================================
# nb_ods — Worker ODS
# Chamado via runMultiple pelo nb_orquestrador
# Lê o parquet mais recente da Landing, aplica flatten + tipagem e grava ODS (Delta)
# =============================================================================
# CÉLULA 1 — Parâmetros (tag esta célula como "parameters" no Fabric)
# =============================================================================

if 'endpoint_alias' not in globals():
    endpoint_alias = ""     # ex: "user", "datasets", "report_audit"
if 'load_type' not in globals():
    load_type = "full"      # full | incremental_datetime
```

```python
# =============================================================================
# CÉLULA 2 — Helpers e configuração Spark
# =============================================================================
```

```python
logger = setup_logger("nb_ods")
configure_spark_ods()
```

```python
# =============================================================================
# CÉLULA 3 — Constantes
# =============================================================================

LAKEHOUSE  = "lh_bronze"
ODS_SCHEMA = "power_embedded"
ODS_TABLE  = f"power_embedded_{endpoint_alias}"

workspace = mssparkutils.env.getWorkspaceName()

landing_folder = build_abfss_path(workspace, LAKEHOUSE, "Files", "landing", "powerembedded", endpoint_alias)
ods_table_path = build_abfss_path(workspace, LAKEHOUSE, "Tables", ODS_SCHEMA, ODS_TABLE)

logger.info(f"endpoint_alias : {endpoint_alias}")
logger.info(f"load_type      : {load_type}")
logger.info(f"landing_folder : {landing_folder}")
logger.info(f"ods_table_path : {ods_table_path}")
```

```python
# =============================================================================
# CÉLULA 4 — Encontrar arquivo mais recente na Landing
# =============================================================================

files = mssparkutils.fs.ls(landing_folder)
parquet_files = [f for f in files if f.name.endswith(".parquet")]

if not parquet_files:
    raise ValueError(f"Nenhum arquivo parquet encontrado em: {landing_folder}")

latest_file = max(parquet_files, key=lambda f: f.modifyTime)
logger.info(f"Arquivo selecionado: {latest_file.name}")
```

```python
# =============================================================================
# CÉLULA 5 — Leitura do parquet (tudo StringType vindo da Landing)
# =============================================================================

df = spark.read.parquet(latest_file.path)
logger.info(f"{df.count()} registro(s) lidos | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 6 — Flatten recursivo de colunas JSON
# =============================================================================

logger.info("Iniciando flatten de colunas JSON...")
df = flatten_json_columns(df, logger=logger, max_depth=5)
logger.info(f"Após flatten: {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 7 — Type casting de colunas escalares
# =============================================================================

logger.info("Iniciando type casting...")
df = cast_string_columns(df, logger=logger, skip_cols=["dt_extracao"])
logger.info("Type casting concluído.")
```

```python
# =============================================================================
# CÉLULA 8 — Adicionar dt_carga
# =============================================================================

from datetime import datetime, timezone
from pyspark.sql import functions as F
from pyspark.sql.types import TimestampType

dt_carga = datetime.now(timezone.utc)
df = df.withColumn("dt_carga", F.lit(dt_carga).cast(TimestampType()))
```

```python
# =============================================================================
# CÉLULA 9 — Gravar ODS
# =============================================================================

if load_type == "full":
    logger.info("Gravando ODS — mode: overwrite")
    (
        df.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .save(ods_table_path)
    )
    logger.info(f"ODS gravado: {ODS_TABLE}")

elif load_type == "incremental_datetime":
    logger.info("Gravando ODS — mode: append")

    # Alinhar schema com a tabela existente antes do append.
    # cast_string_columns usa amostragem e pode inferir tipos diferentes entre execuções
    # (ex: coluna era null na carga inicial → StringType; agora tem valor → LongType).
    # Delta recusa append quando o tipo muda — mesmo com mergeSchema.
    # Solução: castear o novo DF para bater com o schema já gravado.
    try:
        existing_schema = spark.read.format("delta").load(ods_table_path).schema
        for field in existing_schema.fields:
            if field.name in df.columns and df.schema[field.name].dataType != field.dataType:
                logger.info(f"  schema align | {field.name}: {df.schema[field.name].dataType} → {field.dataType}")
                df = df.withColumn(field.name, F.col(field.name).cast(field.dataType))
    except Exception:
        pass  # tabela ainda não existe — primeira carga, sem schema para alinhar

    (
        df.write
        .format("delta")
        .mode("append")
        .option("mergeSchema", "true")
        .save(ods_table_path)
    )
    logger.info(f"ODS atualizado (append): {ODS_TABLE}")

else:
    raise ValueError(f"load_type inválido: {load_type}. Esperado: full | incremental_datetime")
```

```python
# =============================================================================
# CÉLULA 10 — Finalização
# =============================================================================

row_count = df.count()

try:
    log_execucao(logger, "ODS", ODS_TABLE, "success", row_count)
except Exception as e:
    logger.warning(f"Falha ao registrar execução (não crítico): {e}")

result = {
    "status":         "success",
    "endpoint_alias": endpoint_alias,
    "load_type":      load_type,
    "rows":           row_count,
    "columns":        len(df.columns),
    "ods_table":      ODS_TABLE,
}

logger.info(f"nb_ods finalizado | {result}")
mssparkutils.notebook.exit(json.dumps(result))
```
