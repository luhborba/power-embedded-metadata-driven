```python
# =============================================================================
# nb_dw_bridge_datasets_rls — Bridge DW: RLS de Datasets
# Fonte: ods_datasets_rls
# Destino: dw_bridge_datasets_rls
#
# Cada linha = um membro (usuário ou grupo) em uma role RLS de um dataset.
#
# Schema ODS confirmado:
#   id, name, users (array de GUIDs), groups (array de GUIDs),
#   datasetId, userEmails (redundante — descartado), dt_extracao, dt_carga
#
# Join strategy:
#   users[*]  → GUID = sk_user  em dw_dim_user  (join direto por ID)
#   groups[*] → GUID = sk_group em dw_dim_group (join direto por ID)
#   datasetId → nk_dataset_pbi em dw_dim_dataset → sk_dataset
# =============================================================================
# CÉLULA 1 — Parâmetros (tag esta célula como "parameters" no Fabric)
# =============================================================================

if 'dw_table_name'    not in globals(): dw_table_name    = "dw_bridge_datasets_rls"
if 'source_ods_table' not in globals(): source_ods_table = "power_embedded_datasets_rls"
```

```python
logger = setup_logger("nb_dw_bridge_datasets_rls")
configure_spark_dw()
```

```python
# =============================================================================
# CÉLULA 3 — Constantes e caminhos
# =============================================================================

from pyspark.sql import functions as F
from pyspark.sql.types import TimestampType, ArrayType, StringType
from datetime import datetime, timezone

LAKEHOUSE_BRONZE  = "lh_bronze"
LAKEHOUSE_GOLD  = "lh_gold"
ODS_SCHEMA = "power_embedded"
DW_SCHEMA  = "power_embedded"

workspace = mssparkutils.env.getWorkspaceName()

ods_table_path = build_abfss_path(workspace, LAKEHOUSE_BRONZE, "Tables", ODS_SCHEMA, source_ods_table)
dw_table_path  = build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA,  dw_table_name)


logger.info(f"source_ods_table: {source_ods_table}")
logger.info(f"dw_table_name   : {dw_table_name}")
```

```python
# =============================================================================
# CÉLULA 4 — Leitura da ODS
# =============================================================================

df_raw = spark.read.format("delta").load(ods_table_path)
logger.info(f"{df_raw.count()} role(s) RLS lidas | {len(df_raw.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 5 — Remover colunas técnicas + aplicar taxonomia base
# =============================================================================

df_raw = df_raw.drop("dt_extracao", "dt_carga", "userEmails")

df_raw = (
    df_raw
    .withColumnRenamed("id",        "nk_role_rls")
    .withColumnRenamed("name",      "nm_role_rls")
    .withColumnRenamed("datasetId", "nk_dataset_pbi")
)

logger.info("Taxonomia base aplicada.")
```

```python
# =============================================================================
# CÉLULA 6 — Explodir arrays em linhas separadas
#
# users  = ["guid1", "guid2"]  → explode → sk_user  = guid  (ID direto)
# groups = ["guid1", "guid2"]  → explode → sk_group = guid  (ID direto)
# cd_tipo_membro distingue User de Group na linha final.
# =============================================================================

array_str = ArrayType(StringType())

# Linhas de usuários
df_users = (
    df_raw
    .withColumn("users", F.from_json(F.col("users"), array_str))
    .withColumn("sk_user", F.explode_outer(F.col("users")))
    .drop("users", "groups")
    .withColumn("cd_tipo_membro", F.lit("User"))
    .withColumn("sk_group", F.lit(None).cast(StringType()))
)

# Linhas de grupos
df_groups = (
    df_raw
    .withColumn("groups", F.from_json(F.col("groups"), array_str))
    .withColumn("sk_group", F.explode_outer(F.col("groups")))
    .drop("users", "groups")
    .withColumn("cd_tipo_membro", F.lit("Group"))
    .withColumn("sk_user", F.lit(None).cast(StringType()))
)

df = df_users.unionByName(df_groups)

# Remove linhas onde o membro é null (roles sem usuários ou grupos)
df = df.filter(F.col("sk_user").isNotNull() | F.col("sk_group").isNotNull())

logger.info(f"Após explode: {df.count()} linha(s) | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 7 — Enriquecer com sk_dataset via join na dim
#
# sk_user e sk_group já são os GUIDs corretos — sem join adicional necessário.
# Apenas dataset precisa de lookup: datasetId (PBI ID) → sk_dataset (interno).
# Leitura via ABFSS path (não spark.table) para evitar dependência do lakehouse default.
# =============================================================================

dim_dataset = spark.read.format("delta").load(build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA, "dw_dim_dataset")).select("nk_dataset_pbi", "sk_dataset")

df = df.join(dim_dataset, on="nk_dataset_pbi", how="left").drop("nk_dataset_pbi")

nulls_dataset = df.filter(F.col("sk_dataset").isNull()).count()
total = df.count()

logger.info(f"Join cobertura | total: {total} | sk_dataset null: {nulls_dataset}")

if nulls_dataset > 0:
    logger.warning(f"{nulls_dataset} linha(s) sem match em dw_dim_dataset — verificar nk_dataset_pbi")
```

```python
# =============================================================================
# CÉLULA 8 — Deduplicar
# =============================================================================

before = df.count()
df = df.dropDuplicates()
after  = df.count()

if before != after:
    logger.info(f"Deduplicação: {before - after} linha(s) removida(s)")
```

```python
# =============================================================================
# CÉLULA 9 — Adicionar dt_carga
# =============================================================================

dt_carga = datetime.now(timezone.utc)
df = df.withColumn("dt_carga", F.lit(dt_carga).cast(TimestampType()))
```

```python
# =============================================================================
# CÉLULA 10 — Gravar DW
# =============================================================================

logger.info(f"Gravando {dw_table_name} — mode: overwrite")

(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .save(dw_table_path)
)

row_count = df.count()
logger.info(f"DW gravado: {dw_table_name} | {row_count} linha(s) | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 11 — Finalização
# =============================================================================

result = {
    "status":           "success",
    "dw_table_name":    dw_table_name,
    "source_ods_table": source_ods_table,
    "rows":             row_count,
    "columns":          len(df.columns),
}

logger.info(f"nb_dw_bridge_datasets_rls finalizado | {result}")
try:
    log_execucao(logger, "DW", dw_table_name, "success", row_count)
except Exception as e:
    logger.warning(f"Falha ao registrar execução (não crítico): {e}")

mssparkutils.notebook.exit(json.dumps(result))
```
