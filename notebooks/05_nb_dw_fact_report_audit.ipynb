```python
# =============================================================================
# nb_dw_fact_report_audit — Fato DW: Auditoria de Relatórios
# Fonte: ods_report_audit
# Destino: dw_fact_report_audit
#
# Cada linha = um evento de acesso/interação de um usuário em um relatório.
#
# Schema ODS confirmado:
#   id, createdAt, organizationId, userId, userEmail (redundante),
#   reportId, reportName (redundante), action, actionName,
#   dt_extracao, dt_carga
#
# FK strategy:
#   userId  → fk_user   (GUID = sk_user  em dw_dim_user  — join direto por ID)
#   reportId → fk_report (GUID = sk_report em dw_dim_report — join direto por ID)
#   userEmail e reportName descartados (disponíveis via join nas dims)
# =============================================================================
# CÉLULA 1 — Parâmetros (tag esta célula como "parameters" no Fabric)
# =============================================================================

if 'dw_table_name'    not in globals(): dw_table_name    = "dw_fact_report_audit"
if 'source_ods_table' not in globals(): source_ods_table = "power_embedded_report_audit"
```

```python
# =============================================================================
# CÉLULA 2 — Helpers e configuração Spark
# =============================================================================
```

```python
%run 00_nb_helpers
```

```python
logger = setup_logger("nb_dw_fact_report_audit")
configure_spark_dw()
```

```python
# =============================================================================
# CÉLULA 3 — Constantes e caminhos
# =============================================================================

from pyspark.sql import functions as F
from pyspark.sql.types import TimestampType
from datetime import datetime, timezone

LAKEHOUSE_BRONZE  = "lh_bronze"
LAKEHOUSE_GOLD  = "lh_gold"
ODS_SCHEMA = "power_embedded"
DW_SCHEMA  = "power_embedded"

workspace = mssparkutils.env.getWorkspaceName()

ods_table_path = build_abfss_path(workspace, LAKEHOUSE_BRONZE, "Tables", ODS_SCHEMA, source_ods_table)
dw_table_path  = build_abfss_path(workspace, LAKEHOUSE_GOLD, "Tables", DW_SCHEMA,  dw_table_name)

logger.info(f"source_ods_table: {source_ods_table}")
logger.info(f"dw_table_name   : {dw_table_name}")
```

```python
# =============================================================================
# CÉLULA 4 — Leitura da ODS
# =============================================================================

df = spark.read.format("delta").load(ods_table_path)
logger.info(f"{df.count()} evento(s) lidos | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 5 — Aplicar taxonomia + remover redundâncias
# =============================================================================

df = df.drop("dt_extracao", "dt_carga")

# Campos redundantes com as dims — disponíveis via join analítico no Power BI
df = df.drop("userEmail", "reportName")

RENAME_MAP = {
    "id":             "nk_evento",       # natural key do evento de auditoria
    "createdAt":      "dt_evento",       # timestamp do evento
    "organizationId": "cd_organizacao",
    "userId":         "fk_user",         # GUID = sk_user em dw_dim_user
    "reportId":       "fk_report",       # GUID = sk_report em dw_dim_report
    "action":         "cd_acao",         # código numérico da ação
    "actionName":     "nm_acao",         # descrição da ação
}

for old_name, new_name in RENAME_MAP.items():
    if old_name in df.columns:
        df = df.withColumnRenamed(old_name, new_name)

logger.info(f"Taxonomia aplicada | colunas: {df.columns}")
```

```python
# =============================================================================
# CÉLULA 6 — Verificar cobertura referencial
#
# userId e reportId devem bater diretamente com sk_user e sk_report das dims.
# Se houver nulls aqui, o dado veio sem ID na ODS (raro) — investigar.
# =============================================================================

nulls_user   = df.filter(F.col("fk_user").isNull()).count()
nulls_report = df.filter(F.col("fk_report").isNull()).count()
total        = df.count()

logger.info(f"Cobertura FK | total: {total} | fk_user null: {nulls_user} | fk_report null: {nulls_report}")

if nulls_user > 0:
    logger.warning(f"{nulls_user} evento(s) sem fk_user — userId veio null da ODS")
if nulls_report > 0:
    logger.warning(f"{nulls_report} evento(s) sem fk_report — reportId veio null da ODS")
```

```python
# =============================================================================
# CÉLULA 7 — Deduplicar por natural key
# =============================================================================

before = df.count()
df = df.dropDuplicates(["nk_evento"])
after  = df.count()

if before != after:
    logger.info(f"Deduplicação: {before - after} linha(s) removida(s)")
```

```python
# =============================================================================
# CÉLULA 8 — Adicionar dt_carga
# =============================================================================

dt_carga = datetime.now(timezone.utc)
df = df.withColumn("dt_carga", F.lit(dt_carga).cast(TimestampType()))
```

```python
# =============================================================================
# CÉLULA 9 — Gravar DW
# =============================================================================

logger.info(f"Gravando {dw_table_name} — mode: overwrite")

(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .save(dw_table_path)
)

row_count = df.count()
logger.info(f"DW gravado: {dw_table_name} | {row_count} linha(s) | {len(df.columns)} coluna(s)")
```

```python
# =============================================================================
# CÉLULA 10 — Finalização
# =============================================================================

result = {
    "status":           "success",
    "dw_table_name":    dw_table_name,
    "source_ods_table": source_ods_table,
    "rows":             row_count,
    "columns":          len(df.columns),
}

logger.info(f"nb_dw_fact_report_audit finalizado | {result}")
try:
    log_execucao(logger, "DW", dw_table_name, "success", row_count)
except Exception as e:
    logger.warning(f"Falha ao registrar execução (não crítico): {e}")

mssparkutils.notebook.exit(json.dumps(result))
```
