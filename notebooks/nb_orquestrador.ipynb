```python
logger = setup_logger("nb_orquestrador")
```

```python
# =============================================================================
# CÉLULA 3 — Carregar parâmetros do pipeline
# =============================================================================

params = get_pipeline_parameters(logger)
```

```python
# =============================================================================
# CÉLULA 4 — Landing (normal e heavy em sequência por grupo)
# =============================================================================

concurrency_map = {"normal": 5, "heavy": 2}
dags = create_landing_dag(params["endpoint_list"], concurrency_map)

def _check_failures(result: dict, stage: str, logger) -> None:
    """Lança exceção se alguma activity do runMultiple falhou."""
    failed = [name for name, r in result.items() if str(r.get("exitCode", "0")) != "0"]
    if failed:
        raise Exception(f"[{stage}] Falha nas activities: {failed}")

if dags["normal"]["activities"]:
    logger.info(f"Iniciando Landing normal | {len(dags['normal']['activities'])} endpoint(s) | concorrência 5")
    result_normal = mssparkutils.notebook.runMultiple(dags["normal"])
    _check_failures(result_normal, "Landing normal", logger)
    logger.info("Landing normal concluído.")
else:
    logger.info("Landing normal: nenhum endpoint ativo.")

if dags["heavy"]["activities"]:
    logger.info(f"Iniciando Landing heavy | {len(dags['heavy']['activities'])} endpoint(s) | concorrência 2")
    result_heavy = mssparkutils.notebook.runMultiple(dags["heavy"])
    _check_failures(result_heavy, "Landing heavy", logger)
    logger.info("Landing heavy concluído.")
else:
    logger.info("Landing heavy: nenhum endpoint ativo.")
```

```python
# =============================================================================
# CÉLULA 5 — ODS
# =============================================================================

dag_ods = create_ods_dag(params["endpoint_list"])

if dag_ods["activities"]:
    logger.info(f"Iniciando ODS | {len(dag_ods['activities'])} endpoint(s) | concorrência 5")
    try:
        result_ods = mssparkutils.notebook.runMultiple(dag_ods)
    except Exception as e:
        result_ods = e.result if hasattr(e, "result") else {}
    _check_failures(result_ods, "ODS", logger)
    logger.info("ODS concluído.")
else:
    logger.info("ODS: nenhum endpoint ativo.")
```

```python
# =============================================================================
# CÉLULA 6 — DW (dims → bridges → fatos, respeitando dependências do DAG)
# =============================================================================

dag_dw = create_dw_dag(params["dw_list"])

if dag_dw["activities"]:
    logger.info(f"Iniciando DW | {len(dag_dw['activities'])} tabela(s) | concorrência 5")
    try:
        result_dw = mssparkutils.notebook.runMultiple(dag_dw)
    except Exception as e:
        result_dw = e.result if hasattr(e, "result") else {}
    _check_failures(result_dw, "DW", logger)
    logger.info("DW concluído.")
else:
    logger.info("DW: nenhuma tabela na lista.")
```

```python
# =============================================================================
# CÉLULA 7 — Encerramento
# =============================================================================

result = {
    "status":            "success",
    "endpoints_landing": len(params["endpoint_list"]),
    "endpoints_ods":     len(dag_ods["activities"]),
    "tables_dw":         len(dag_dw["activities"]),
}

logger.info(f"nb_orquestrador finalizado | {result}")
mssparkutils.notebook.exit(result)
```
