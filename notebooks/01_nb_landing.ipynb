```python
# =============================================================================
# nb_landing — Worker de extração Landing
# Recebe parâmetros de um endpoint e salva Parquet bruto no Lakehouse
# =============================================================================
# CÉLULA 1 — Parâmetros (tag esta célula como "parameters" no Fabric)
# =============================================================================

if 'endpoint'          not in globals(): endpoint          = ""
if 'endpoint_alias'    not in globals(): endpoint_alias    = ""
if 'is_paginated'      not in globals(): is_paginated      = 1
if 'load_type'         not in globals(): load_type         = "full"
if 'delta_param_start' not in globals(): delta_param_start = ""
if 'delta_param_end'   not in globals(): delta_param_end   = ""
if 'last_delta_value'  not in globals(): last_delta_value  = ""
```

```python
endpoint          = "/api/capacity-audit"
endpoint_alias    = "capacity_audit"
is_paginated      = 1
load_type         = "full"
delta_param_start = ""
delta_param_end   = ""
last_delta_value  = ""
```

```python
# =============================================================================
# CÉLULA 2 — Imports e helpers
# =============================================================================
```

```python
import requests
from datetime import datetime, timezone, timedelta
from pyspark.sql import functions as F
from pyspark.sql.types import TimestampType

logger = setup_logger("nb_landing")
```

```python
# =============================================================================
# CÉLULA 3 — Constantes
# =============================================================================

BASE_URL         = "https://api.powerembedded.com.br"
LANDING_BASE     = "Files/landing/powerembedded"
RETENTION_DAYS   = 7
```

```python
# =============================================================================
# CÉLULA 4 — Validação de parâmetros
# =============================================================================

if not endpoint:
    raise ValueError("Parâmetro 'endpoint' não informado.")
if not endpoint_alias:
    raise ValueError("Parâmetro 'endpoint_alias' não informado.")

logger.info(f"Iniciando landing | alias={endpoint_alias} | load_type={load_type}")
```

```python
# =============================================================================
# CÉLULA 5 — Obter API key
# =============================================================================

key = get_api_key(logger)
headers = {"X-API-Key": key}
```

```python
# =============================================================================
# CÉLULA 6 — Montar parâmetros de query (incremental vs full)
# =============================================================================

query_params = {}

if load_type == "incremental_datetime":
    start = last_delta_value if last_delta_value else "2024-01-01T00:00:00"
    end   = datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%S")

    if not delta_param_start or not delta_param_end:
        raise ValueError(
            f"load_type=incremental_datetime exige delta_param_start e delta_param_end. "
            f"Verifique a tabela de controle para '{endpoint_alias}'."
        )

    query_params[delta_param_start] = start
    query_params[delta_param_end]   = end
    logger.info(f"Incremental | {delta_param_start}={start} | {delta_param_end}={end}")
else:
    end = None
```

```python
# =============================================================================
# CÉLULA 7 — Extração com paginação
# =============================================================================

all_records = []

if is_paginated:
    page = 1
    while True:
        params = {**query_params, "pageNumber": page}
        url    = f"{BASE_URL}{endpoint}"

        logger.info(f"GET {url} | página {page}")
        response = requests.get(url, headers=headers, params=params, timeout=300)
        response.raise_for_status()

        body = response.json()
        data = body.get("data", [])
        all_records.extend(data)

        logger.info(f"Página {page} | {len(data)} registro(s) | total acumulado: {len(all_records)}")

        if not body.get("hasNextPage", False):
            break

        page += 1
else:
    url      = f"{BASE_URL}{endpoint}"
    logger.info(f"GET {url} | não paginado")
    response = requests.get(url, headers=headers, params=query_params, timeout=300)
    response.raise_for_status()

    body = response.json()
    # Endpoint não paginado pode retornar lista direta ou envelope com "data"
    all_records = body if isinstance(body, list) else body.get("data", body)

logger.info(f"Extração concluída | total: {len(all_records)} registro(s)")
```

```python
# =============================================================================
# CÉLULA 8 — Verificar se há registros
# =============================================================================

if not all_records:
    logger.warning(f"Nenhum registro retornado para '{endpoint_alias}'. Encerrando sem salvar arquivo.")
    mssparkutils.notebook.exit(json.dumps({
        "status":         "empty",
        "endpoint_alias": endpoint_alias,
        "records":        0,
        "file_path":      None,
    }))
```

```python
# =============================================================================
# CÉLULA 9 — Criar DataFrame e adicionar dt_extracao
# =============================================================================

df = records_to_landing_df(all_records)
dt_extracao = datetime.now(timezone.utc)
df = df.withColumn("dt_extracao", F.lit(dt_extracao).cast(TimestampType()))

logger.info(f"DataFrame criado | {df.count()} linhas | {len(df.columns)} colunas")
```

```python
# =============================================================================
# CÉLULA 10 — Salvar Parquet na Landing
# =============================================================================

ts        = dt_extracao.strftime("%Y%m%d_%H%M%S")
file_name = f"{endpoint_alias}_{ts}.parquet"
file_path = f"{LANDING_BASE}/{endpoint_alias}/{file_name}"

df.write.mode("overwrite").parquet(file_path)
logger.info(f"Parquet salvo | {file_path}")
```

```python
# =============================================================================
# CÉLULA 11 — Limpeza de retenção (arquivos com mais de 7 dias)
# =============================================================================

folder_path  = f"{LANDING_BASE}/{endpoint_alias}"
cutoff_ms    = (dt_extracao - timedelta(days=RETENTION_DAYS)).timestamp() * 1000

try:
    files = notebookutils.fs.ls(folder_path)
    deleted = 0
    for f in files:
        if f.modifyTime < cutoff_ms:
            notebookutils.fs.rm(f.path, recurse=False)
            logger.info(f"Arquivo removido (retenção): {f.path}")
            deleted += 1
    logger.info(f"Limpeza concluída | {deleted} arquivo(s) removido(s)")
except Exception as e:
    logger.warning(f"Erro na limpeza de retenção (não crítico): {e}")
```

```python
# =============================================================================
# CÉLULA 12 — Atualizar last_delta_value (somente incremental_datetime)
# =============================================================================

if load_type == "incremental_datetime" and end:
    try:
        conn   = get_db_connection(logger)
        cursor = conn.cursor()
        cursor.execute(
            "UPDATE powerembedded_control SET last_delta_value = ? WHERE endpoint_alias = ?",
            end, endpoint_alias
        )
        conn.commit()
        conn.close()
        logger.info(f"last_delta_value atualizado | endpoint_alias={endpoint_alias} | valor={end}")
    except Exception as e:
        logger.error(f"Falha ao atualizar last_delta_value para '{endpoint_alias}': {e}")
        raise
```

```python
# =============================================================================
# CÉLULA 13 — Retorno
# =============================================================================

result = {
    "status":         "success",
    "endpoint_alias": endpoint_alias,
    "records":        len(all_records),
    "file_path":      file_path,
}

logger.info(f"nb_landing finalizado | {result}")
mssparkutils.notebook.exit(json.dumps(result))
```
